### *Worst* scoring samples from testing dataset, for four types of IoU scores: high density, medium density, low density, and overall/combined.


```python
for cat in [
            'high', 
            'medium',
            'low', 
            'overall'
           ]:
    print(cat.upper() + ' DENSITY')
    plot_cases(category=cat, iou_type='worst', n=5)
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
import torch
import os
import skimage.io
import json
from glob import glob
import numpy as np
from datetime import datetime

def score_to_color(score):
    if np.isnan(score):
        return (0.8, 0.8, 0.8) 
    red = max(0, 1 - score)     # Linearly interpolate between red (1, 0, 0) and green (0, 1, 0)
    green = min(1, score)
    return (red, green, 0)  # RGB color tuple

def plot_cases(category, iou_type, fill_labels=True, RGB_HMS_model=True, n=20):
    # category could be 'low','medium', 'high', 'overall'
    # iou type could be 'best' or 'worst'
    # test_results_1731541247
    save_dir = os.path.join(os.getcwd(), f'test_results_1731541247/{category}/{iou_type}/')
    case_dirs = [os.path.join(save_dir, case) for case in os.listdir(save_dir)]    
    # colors = ['red', 'orange', 'yellow']
    colors = [(1, 0, 0), (1, 0.65, 0), (1, 1, 0)]
    colors_with_alpha = [(r, g, b, 0.25) for r, g, b in colors]  
    labels = ['high', 'medium', 'low']

    for i, case_dir in enumerate(case_dirs):
        fig, ax = plt.subplots(1, 3, figsize=(15, 7))
        # Load predictions and fn_info
        preds = skimage.io.imread(os.path.join(case_dir, 'preds.tif'))
        with open(os.path.join(case_dir, 'fn_info.json'), 'r') as f:
            fn_info = json.load(f)

        # Extract IoU scores
        low_iou = float(fn_info['low_iou'])
        medium_iou = float(fn_info['medium_iou'])
        high_iou = float(fn_info['high_iou'])
        overall_iou = float(fn_info['overall_iou'])
        iou_scores = { "Low IoU": low_iou, "Medium IoU": medium_iou, "High IoU": high_iou, "Overall IoU": overall_iou }
        
        fn = os.path.basename(fn_info['truth_fn'])
        data_loc = '/scratch1/RDARCH/rda-ghpcs/Rey.Koki/SmokeViz/'
        data_fn = glob(data_loc + "data/*/*/" + fn)[0]
        truth_fn = glob(data_loc + "truth/*/*/" + fn)[0]
        print(truth_fn)
        coords_fn = glob(data_loc + "coords/*/*/" + fn)[0]
        RGB = skimage.io.imread(data_fn, plugin='tifffile')
        truths = skimage.io.imread(truth_fn, plugin='tifffile')
        lat_lon = skimage.io.imread(coords_fn, plugin='tifffile')
        lat = lat_lon[:, :, 0]; lon = lat_lon[:, :, 1]
        num_pixels = RGB.shape[1]
        x = np.linspace(0,num_pixels-1,num_pixels)
        y = np.linspace(0,num_pixels-1,num_pixels)
        X, Y = np.meshgrid(x,y)
        start = fn.split('_')[1][1:-1]
        start_dt = datetime.strptime(start, '%Y%j%H%M%S')
        start_readable = start_dt.strftime('%Y/%m/%d %H:%M UTC')
        # plt.suptitle(start_readable + f' (selected in 20 {iou_type}-scoring for {category} density)', fontsize=15)
        print(case_dir, start_readable)
        print(f'selected in 20 {iou_type}-scoring for {category} density')

        ax[0].imshow(RGB); ax[1].imshow(RGB); ax[2].imshow(RGB)
        preds = torch.sigmoid(torch.tensor(preds))
        preds = (preds > 0.5) * 1

        legend_list = [Patch(color=colors[idx], label=labels[idx]) for idx in range(3)]  
        fig.legend(handles=legend_list, loc='upper right', bbox_to_anchor=(0.95, 0.95), 
                   title="plume density", fontsize=10, title_fontsize=12, ncols=3)
        im_shape = truths[:, :, 0].shape
        for idx, color in list(enumerate(colors_with_alpha)):
            curr_preds = preds.squeeze()[:, :, idx].numpy()
            curr_truths = truths[:, :, idx]
            # print(curr_preds.min(), curr_preds.max(), curr_preds.shape)
            # print(curr_truths.min(), curr_truths.max(), curr_truths.shape)
            truth_overlay = np.zeros((*im_shape, 4)) 
            preds_overlay = np.zeros((*im_shape, 4)) 
            truth_overlay[curr_truths > 0.99] = color  
            preds_overlay[curr_preds> 0.99] = color 
            
            # fill in pixels
            ax[1].imshow(truth_overlay)  
            ax[2].imshow(preds_overlay)  
            # draw contours
            ax[1].contour(X, Y, curr_truths, levels=[0.99], colors=[color[:-1]])
            ax[2].contour(X, Y, curr_preds, levels=[0.99], colors=[color[:-1]])
            # ax[2].contour(X, Y, preds.squeeze()[:, :, idx], levels=[0.99], colors=[color[:-1]]) # this is wrong since doesn't convert pred to class
            
        ax[0].set_title(start_readable)
        ax[1].set_title('HMS Analyst Labels')
        ax[2].set_title('Model Predictions')
        ax[2].set_xticklabels([]); ax[2].set_yticks([]); ax[2].set_yticklabels([]); ax[2].set_xticks([])
            
        num_ticks = 5
        ytick_locations = np.linspace(0, 255, num_ticks)
        xtick_locations = np.linspace(0, 255, num_ticks)
        ytick_labels = np.round(np.linspace(lat.min(), lat.max(), num_ticks), 2)
        xtick_labels = np.round(np.linspace(lon.min(), lon.max(), num_ticks), 2)
        ax[0].set_yticks(ytick_locations); ax[0].set_yticklabels(ytick_labels, fontsize=12)
        ax[0].set_xticks(xtick_locations); ax[0].set_xticklabels(xtick_labels, fontsize=12)
        ax[1].set_yticklabels([]); ax[1].set_xticklabels([])
        ax[1].set_yticks([]); ax[1].set_xticks([])
        for x_pos, (label, score) in zip([0.1, 0.2, 0.3, 0.4], iou_scores.items()):
            color = score_to_color(score)
            fig.text(x_pos, 0.9, f"{label}: {score:.3f}" if not np.isnan(score) else f"{label}: NaN",
                     ha='center', fontsize=10, verticalalignment='center', 
                     bbox=dict(facecolor=color, alpha=0.6, boxstyle="round,pad=0.5"))

        plt.tight_layout(pad=2)
        plt.show()

        if i > n - 2:
            break

import warnings
warnings.filterwarnings('ignore')
```
