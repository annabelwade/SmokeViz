# **Learning Curves**
from https://machinelearningmastery.com/learning-curves-for-diagnosing-machine-learning-model-performance/:
- overfitting often occurs if the model has more capacity than is required for the problem, and, in turn, too much flexibility. It can also occur if the model is trained for too long.
- A plot of learning curves shows overfitting if:
    - The plot of training loss continues to decrease with experience.
    - The plot of validation loss decreases to a point and begins increasing agai
- A plot of learning curves shows a good fit if:
    - The plot of training loss decreases to a point of stability.
    - The plot of validation loss decreases to a point of stability and has a small gap with the training loss.

```python
def learning_curves(list_of_history_dicts, suptitle=None, subplot_titles=None, num_epochs=40,):
    num_experiments = len(list_of_history_dicts)
    fig, axs = plt.subplots(1, num_experiments, figsize=(num_experiments * 6 + 5, 6),
                            sharex=False,
                            sharey=False)

    if num_experiments == 1:
        axs = np.expand_dims(axs, axis=0)

    for i in range(num_experiments):
        # Get values for each experiment
        history = list_of_history_dicts[i]
        train_loss = history['train_loss']
        val_loss = history['val_loss']
        val_iou = history['val_iou']

        train_loss = train_loss[:num_epochs]
        val_loss = val_loss[:num_epochs]
        val_iou = val_iou[:num_epochs]

        # Calculate the epoch numbers for plotting
        epoch_nums = np.arange(1, len(train_loss) + 1)

        # Plot training loss in the first row
        axs[i].plot(epoch_nums, train_loss, label='Training Loss', marker='o')
        axs[i].plot(epoch_nums, val_loss, label='Validation Loss', marker='o', color='orange')
        # axs[i].set_title('Experiment {}'.format(i+1))
        axs[i].set_xlabel('Epoch')
        axs[i].set_ylabel('Loss')
        axs[i].grid(True, alpha=0.2)

        # Create a second y-axis for the validation IoU
        ax2 = axs[i].twinx()
        ax2.plot(epoch_nums, val_iou, label='Validation IoU', marker='o', color='darkgreen')
        ax2.set_ylabel('Combined IoU', color='darkgreen')
        ax2.tick_params(axis='y', colors='darkgreen')
        ax2.grid(True, alpha=0.2)

        # Add legend to the plot
        if i == num_experiments - 1:
            lines, labels = axs[i].get_legend_handles_labels()
            lines2, labels2 = ax2.get_legend_handles_labels()
            # place the legend outside of the plot.
            axs[i].legend(lines + lines2, labels + labels2, loc='upper left', bbox_to_anchor=(1.07, 1), fontsize=18)
    
    if suptitle:
        fig.suptitle(suptitle, fontsize=20)
    if subplot_titles:
        for ind, title in enumerate(subplot_titles):
            axs[ind].set_title(title, fontsize=20)

    plt.tight_layout()
    plt.show()
```

```python
# PRECISION AND RECALL
nan = float('nan')
exp3_seed20_history_precision_recall = {'train_loss': [1.2389514883500432, 1.098791884116127, 1.018984439741443, 0.9719240083152405, 0.9354802797255677, 0.9070093708499922, 0.8870192515278371, 0.8734155551366654, 0.8633751828441824, 0.8501807128874913, 0.8383572154527871, 0.8325524524410072, 0.8206735237982938, 0.8149967169253484, 0.8098911746251223, 0.8069924067963927, 0.8007039385938729, 0.8004698230625258, 0.7911360862517653, 0.7871647435860778, 0.7849578845776207, 0.7785087968644622, 0.7743459855790248, 0.7706657904291238, 0.7675656929321018, 0.7611530972098796, 0.7613940560796545, 0.7565308411739647, 0.7552835822211278, 0.7500860479331144, 0.7437178292187662, 0.742974644602933, 0.7376788683090074, 0.7371605844046676, 0.734250877535576, 0.7324518082832147, 0.7281868625068326, 0.7291079430342992, 0.7229019860285425, 0.7234228170521517], 'val_loss': [1.7975916790962219, 1.6088696241378784, 1.5104190973440805, 1.4156932246685028, 1.709660683075587, 1.6906954276561736, 1.4862589655319849, 1.3061978509028753, 1.5130251799027126, 1.5096661732594172, 1.4057061590751012, 2.088355552752813, 1.9745376245180766, 1.3419107568264008, 1.3683943873643876, 1.2393536736567816, 1.7003869251410166, 1.1920035431782405, 1.6833463271458944, 1.4186899056037268, 2.210772840976715, 2.036206177075704, 1.3017872460683186, 1.4282058946291605, 1.4434940242767333, 2.4006263701121013, 1.3185365972916285, 1.392266013622284, 1.6765294496218364, 1.5266235025723776, 1.3449221581220627, 2.4002571709950766, 5.942988740603129, 4.498541170358658, 1.3470390003919601, 1.8413161971171697, 1.7503697268168132, 1.6761245661973954, 1.7390277846654256, 2.1603815909226736], 'val_iou': [0.9704528450965881, 1.2880828380584717, 2.8306617736816406, 2.4153144359588623, 2.4550833702087402, 2.3428027629852295, 2.702596664428711, 2.4191439151763916, 2.7373127937316895, 2.4712343215942383, 2.1966352462768555, 2.1291146278381348, 2.032219648361206, 2.604799270629883, 2.750265121459961, 3.1181654930114746, 1.9849255084991455, 2.9932310581207275, 2.6359446048736572, 3.1908891201019287, 2.094757556915283, 2.577481985092163, 3.122910976409912, 2.31819486618042, 3.038752555847168, 2.4684953689575195, 2.7335662841796875, 2.8615686893463135, 2.776073932647705, 1.8437471389770508, 2.875594139099121, 2.535011053085327, 2.2997243404388428, 2.490774154663086, 2.638148069381714, 2.9222402572631836, 2.6884701251983643, 2.5107779502868652, 1.9631404876708984, 2.9120447635650635], 'val_precision': [[nan, 0.8011293411254883, 0.9075304865837097, 0.88515704870224], [0.7617732286453247, 0.8301591873168945, 0.8702133297920227, 0.8628571629524231], [0.6346790194511414, 0.6765970587730408, 0.8314706683158875, 0.773668646812439], [0.7223312854766846, 0.8159213066101074, 0.9192057251930237, 0.8809200525283813], [0.645308256149292, 0.8651383519172668, 0.9314365983009338, 0.8816672563552856], [0.6541042923927307, 0.8854772448539734, 0.9382745623588562, 0.8933752179145813], [0.6722463965415955, 0.818137526512146, 0.8274742364883423, 0.8151395916938782], [0.825395405292511, 0.8641423583030701, 0.8870538473129272, 0.8796110153198242], [0.6657408475875854, 0.8299691677093506, 0.9275414943695068, 0.8747407793998718], [0.8435346484184265, 0.8961738348007202, 0.9123393893241882, 0.9045416116714478], [0.7986789345741272, 0.8666803240776062, 0.8557454347610474, 0.8555092215538025], [0.8302828669548035, 0.9036416411399841, 0.9200053215026855, 0.9116623401641846], [0.8230393528938293, 0.9356191754341125, 0.9162960648536682, 0.9136620759963989], [0.805671751499176, 0.8876387476921082, 0.9211856722831726, 0.9063133001327515], [0.7892840504646301, 0.847998321056366, 0.8969157934188843, 0.8794191479682922], [0.7369481325149536, 0.816532552242279, 0.8558037877082825, 0.8386548161506653], [0.730681836605072, 0.8587396740913391, 0.877513587474823, 0.8666917681694031], [0.7561982870101929, 0.7976142168045044, 0.8228060007095337, 0.813772976398468], [0.7492814660072327, 0.8947553038597107, 0.9250936508178711, 0.9014750123023987], [0.5376964211463928, 0.6524174213409424, 0.7529692649841309, 0.7069485187530518], [0.8212122321128845, 0.916312038898468, 0.937716543674469, 0.9229024648666382], [0.36389997601509094, 0.6034337282180786, 0.795087456703186, 0.6861029267311096], [0.6185768246650696, 0.7379826903343201, 0.8763965964317322, 0.8156328201293945], [0.8636898994445801, 0.9222742915153503, 0.8913703560829163, 0.895138680934906], [0.5769172310829163, 0.7476016879081726, 0.8951864242553711, 0.8192242383956909], [0.7756375074386597, 0.9038043022155762, 0.9217377305030823, 0.9031615257263184], [0.764632523059845, 0.8277872800827026, 0.9003840684890747, 0.8743407726287842], [0.6939080357551575, 0.7115769982337952, 0.8128669857978821, 0.7796590328216553], [0.5933753252029419, 0.7973812818527222, 0.8921878337860107, 0.8335588574409485], [0.9199745655059814, 0.9275586009025574, 0.9134227633476257, 0.915980339050293], [0.797219455242157, 0.903998076915741, 0.8798384070396423, 0.8788435459136963], [0.41183507442474365, 0.6697211265563965, 0.8170599937438965, 0.7198842763900757], [0.43896013498306274, 0.7804471850395203, 0.8594127297401428, 0.7399879693984985], [0.5100141763687134, 0.8075584769248962, 0.7227303385734558, 0.7074121832847595], [0.8052958846092224, 0.8893414735794067, 0.873406708240509, 0.8730207085609436], [0.47205668687820435, 0.7276406288146973, 0.8925598859786987, 0.7855751514434814], [0.41670864820480347, 0.5883846879005432, 0.8448999524116516, 0.7151131629943848], [0.7016247510910034, 0.8737651705741882, 0.9202333092689514, 0.8881211280822754], [0.8365405201911926, 0.9239820241928101, 0.9267839789390564, 0.9215942621231079], [0.4854883551597595, 0.7857457399368286, 0.9124438762664795, 0.80470210313797]], 'val_recall': [[0.0, 0.28396421670913696, 0.4605664610862732, 0.3779098391532898], [0.00602532085031271, 0.36183375120162964, 0.6544504165649414, 0.5269400477409363], [0.4993423521518707, 0.6827610731124878, 0.7321497797966003, 0.7004690766334534], [0.33688539266586304, 0.5290616750717163, 0.6058743000030518, 0.5642682909965515], [0.4927397668361664, 0.44321075081825256, 0.4790765941143036, 0.4711177945137024], [0.4417269825935364, 0.43949294090270996, 0.4497697651386261, 0.44650453329086304], [0.44756659865379333, 0.5365307331085205, 0.7407982349395752, 0.6649146676063538], [0.2792419195175171, 0.5322042107582092, 0.7038019299507141, 0.6253779530525208], [0.5040053129196167, 0.5607678294181824, 0.5471041798591614, 0.5470156669616699], [0.35513731837272644, 0.48908546566963196, 0.5761747360229492, 0.5359147191047668], [0.2634950578212738, 0.43230947852134705, 0.7232791781425476, 0.6117205023765564], [0.2869756817817688, 0.4177168607711792, 0.5403298139572144, 0.4884112775325775], [0.32614830136299133, 0.3565720319747925, 0.43873265385627747, 0.40865203738212585], [0.3817814886569977, 0.5143250226974487, 0.6216525435447693, 0.5747157335281372], [0.39586585760116577, 0.5673514604568481, 0.6980688571929932, 0.6400749683380127], [0.5166949033737183, 0.6441646814346313, 0.7731174230575562, 0.7193405032157898], [0.2666383683681488, 0.37067142128944397, 0.6049590706825256, 0.5177590847015381], [0.41955646872520447, 0.6871713995933533, 0.8440741300582886, 0.7693755030632019], [0.4853236973285675, 0.4611850678920746, 0.5303080081939697, 0.5090963840484619], [0.7434060573577881, 0.8144351840019226, 0.8991643190383911, 0.8648777604103088], [0.34300696849823, 0.3868103325366974, 0.4003901183605194, 0.3922251760959625], [0.6466994881629944, 0.746871292591095, 0.8023567795753479, 0.7754852175712585], [0.6032186150550842, 0.7031429409980774, 0.7503332495689392, 0.7262662053108215], [0.3239296078681946, 0.39071160554885864, 0.6907979249954224, 0.5845818519592285], [0.6290625929832458, 0.6874569058418274, 0.681817889213562, 0.6789016127586365], [0.4391925036907196, 0.46008655428886414, 0.43791964650154114, 0.44363874197006226], [0.3924928903579712, 0.5903453230857849, 0.6788972020149231, 0.6328837871551514], [0.40421053767204285, 0.7453084588050842, 0.8169856071472168, 0.7648395299911499], [0.5959063768386841, 0.5400579571723938, 0.5971203446388245, 0.5825678110122681], [0.15350908041000366, 0.3848772346973419, 0.6835934519767761, 0.5642834901809692], [0.47334787249565125, 0.48585933446884155, 0.754614531993866, 0.6633831858634949], [0.6813919544219971, 0.5901595950126648, 0.6728683114051819, 0.6526220440864563], [0.6952743530273438, 0.48857489228248596, 0.35581475496292114, 0.41739410161972046], [0.6795896291732788, 0.5023112893104553, 0.41782480478286743, 0.46077945828437805], [0.3758256435394287, 0.5157932043075562, 0.699629008769989, 0.626402735710144], [0.7360793948173523, 0.6802515983581543, 0.677804708480835, 0.683223307132721], [0.6432241201400757, 0.7599688172340393, 0.7778940200805664, 0.7622638940811157], [0.4789927303791046, 0.4305572509765625, 0.5295270085334778, 0.5002988576889038], [0.26776108145713806, 0.35029900074005127, 0.5409971475601196, 0.47019699215888977], [0.7434418201446533, 0.6669028401374817, 0.5650673508644104, 0.6055488586425781]]}
exp1_seed20_history_precision_recall = {'train_loss': [1.199456359575739, 1.0563687454002577, 0.990011724919663, 0.9482787875045045, 0.9280259035937951, 0.9074834355981904, 0.8917696301297651, 0.8788936087335617, 0.8695127473313685, 0.8625243889502056, 0.8554640282110888, 0.8480943924156853, 0.8399588017332406, 0.833223102941818, 0.8265788406927674, 0.8243536466921202, 0.8176672450179105, 0.8141410373032198, 0.8093621713335942, 0.8050675747343744, 0.8016176575771657, 0.7965385738634513, 0.7944993970660926, 0.7890014659553819, 0.7825532665155286, 0.7810605876610927, 0.7757996214325432, 0.775230200898372, 0.7702656328995741, 0.7683845476560017, 0.7634259697387316, 0.7591697245465714, 0.7544807465631204, 0.7516518582459242, 0.7459538290492171, 0.7450973155708127, 0.7384257489686326, 0.7341080729321943, 0.7309672637688753, 0.7291072268786692], 'val_loss': [1.7332839542627334, 1.7604453668991724, 1.4231130403280259, 1.4270851439237595, 1.6670041817426682, 1.5173834238449733, 1.4637731603781383, 1.3833091795444488, 1.4378845010201136, 1.3288245209058125, 1.357443779706955, 1.2401464021205901, 1.3330238397916159, 1.2670331525802612, 1.2420347315073013, 1.5003586532672246, 1.5574768843253453, 1.3900241678953171, 1.3121939434607823, 1.247284372051557, 1.4472573826710382, 1.3150270116329192, 1.469398808479309, 1.3669174935420354, 1.3261648386716842, 1.313452224334081, 1.3343188353379567, 1.220637750029564, 1.3389205718040467, 1.2801832962036133, 1.2749717134237288, 1.3762793391942978, 1.248157155116399, 1.3182563422123592, 1.5246756770213445, 1.2707699199517568, 1.2116674651702246, 1.3737308039267857, 1.2763348269462584, 1.271015269358953], 'val_iou': [1.641945719718933, 0.8834288716316223, 2.3034563064575195, 1.748613715171814, 1.5968526601791382, 1.6629154682159424, 1.9208383560180664, 2.385806083679199, 2.637145757675171, 2.915174961090088, 2.1982815265655518, 3.1481845378875732, 2.6515796184539795, 2.764824867248535, 2.8198885917663574, 2.0188398361206055, 1.7007267475128174, 2.3231616020202637, 2.838651657104492, 3.1149823665618896, 1.9629513025283813, 2.8563523292541504, 2.4099252223968506, 2.338222026824951, 2.937910795211792, 2.812617540359497, 2.50144100189209, 2.779545307159424, 2.422549247741699, 2.5810720920562744, 2.947787046432495, 2.817584276199341, 3.157644271850586, 2.641975164413452, 2.1733458042144775, 2.789091110229492, 3.160344123840332, 2.714998722076416, 2.917048931121826, 2.749314308166504], 'val_precision': [[0.6188645958900452, 0.8402228951454163, 0.9354856610298157, 0.8860698342323303], [0.5444796085357666, 0.8773177862167358, 0.8938328623771667, 0.8920018672943115], [0.7240954041481018, 0.8275209665298462, 0.8788982033729553, 0.8608313202857971], [0.8467437028884888, 0.8773455619812012, 0.8595221638679504, 0.8620728850364685], [0.8289489150047302, 0.8878876566886902, 0.9160245656967163, 0.9075000286102295], [0.7718795537948608, 0.7679270505905151, 0.837687611579895, 0.8232100605964661], [0.7846217751502991, 0.8668850064277649, 0.9256137609481812, 0.9073096513748169], [0.7442994713783264, 0.8196115493774414, 0.9212920665740967, 0.8858736753463745], [0.6714606285095215, 0.8443715572357178, 0.9138278961181641, 0.8761336803436279], [0.6579279899597168, 0.8475320935249329, 0.9284744262695312, 0.8805130124092102], [0.7698802947998047, 0.8175451755523682, 0.8564907312393188, 0.8460855484008789], [0.6296209096908569, 0.8372774720191956, 0.9105131030082703, 0.8631071448326111], [0.5848766565322876, 0.7444717288017273, 0.8771178722381592, 0.8237020373344421], [0.7284719347953796, 0.8471347093582153, 0.9013352394104004, 0.8777414560317993], [0.8029614090919495, 0.8663235306739807, 0.9114604592323303, 0.8944668173789978], [0.8805117607116699, 0.9149161577224731, 0.8950245976448059, 0.8982709050178528], [0.8235378265380859, 0.8737474083900452, 0.9131770133972168, 0.9038088917732239], [0.7853189706802368, 0.8765421509742737, 0.9179404377937317, 0.9021084308624268], [0.7038999795913696, 0.8584384322166443, 0.9164227247238159, 0.8852051496505737], [0.6930363178253174, 0.8199149966239929, 0.914409339427948, 0.8707849383354187], [0.9113718867301941, 0.8929808139801025, 0.9310085773468018, 0.922026515007019], [0.7052484750747681, 0.8195074796676636, 0.8949941396713257, 0.8630731701850891], [0.7847097516059875, 0.885306715965271, 0.932876706123352, 0.912128210067749], [0.8126124739646912, 0.8868522047996521, 0.9165434837341309, 0.9053226709365845], [0.7084594964981079, 0.8388956785202026, 0.8856148719787598, 0.8597279191017151], [0.7472768425941467, 0.8469898700714111, 0.9076127409934998, 0.8831952810287476], [0.8302836418151855, 0.8947850465774536, 0.9019511342048645, 0.8968172073364258], [0.7899946570396423, 0.8588176369667053, 0.8799953460693359, 0.8708441853523254], [0.8595280647277832, 0.8904017806053162, 0.9109162092208862, 0.9042256474494934], [0.7628788948059082, 0.8113647103309631, 0.8789387941360474, 0.8587362766265869], [0.7178219556808472, 0.8530471920967102, 0.9132251143455505, 0.8839920163154602], [0.7040414810180664, 0.850825846195221, 0.9213482737541199, 0.8871552348136902], [0.6588729619979858, 0.8218600153923035, 0.9101356863975525, 0.8648279309272766], [0.7751824855804443, 0.8652457594871521, 0.9008123874664307, 0.8855714201927185], [0.8262262940406799, 0.913910984992981, 0.9189727306365967, 0.9132871031761169], [0.7909770011901855, 0.8636502623558044, 0.9015611410140991, 0.8870394229888916], [0.6615918874740601, 0.7723426818847656, 0.880072832107544, 0.83405601978302], [0.7521514296531677, 0.8608296513557434, 0.9115836024284363, 0.8894093632698059], [0.736443042755127, 0.8456887006759644, 0.9112600088119507, 0.8829084634780884], [0.7493884563446045, 0.8445732593536377, 0.8977842926979065, 0.8771473169326782]], 'val_recall': [[0.26230132579803467, 0.3105906546115875, 0.3880559206008911, 0.3580799698829651], [0.0010261228308081627, 0.17326484620571136, 0.5793173313140869, 0.42885199189186096], [0.27235084772109985, 0.514630138874054, 0.6942880749702454, 0.6140386462211609], [0.12452687323093414, 0.3769354522228241, 0.7487766742706299, 0.6031914949417114], [0.17302466928958893, 0.3182011544704437, 0.5063104629516602, 0.4312209486961365], [0.09888800233602524, 0.42070746421813965, 0.7152258157730103, 0.5898761749267578], [0.17195111513137817, 0.46502551436424255, 0.5865775942802429, 0.5216469168663025], [0.2636733055114746, 0.5961460471153259, 0.6396288275718689, 0.5976557731628418], [0.4434530735015869, 0.5396522283554077, 0.5938019156455994, 0.5677059292793274], [0.5489858984947205, 0.5794053077697754, 0.6120588183403015, 0.5985944271087646], [0.22237849235534668, 0.4956771731376648, 0.7675865888595581, 0.6538207530975342], [0.6326294541358948, 0.6226245164871216, 0.6991550922393799, 0.6742933392524719], [0.4000168442726135, 0.6302406787872314, 0.7523273825645447, 0.6923930048942566], [0.42235267162323, 0.5594615936279297, 0.7041794061660767, 0.6443175077438354], [0.3949035406112671, 0.5947248339653015, 0.6942288279533386, 0.6443774700164795], [0.24938789010047913, 0.3913555443286896, 0.5764479041099548, 0.5026351809501648], [0.15525390207767487, 0.3462664484977722, 0.6255426406860352, 0.5160806179046631], [0.28705862164497375, 0.4977589249610901, 0.6277014017105103, 0.5667382478713989], [0.49941831827163696, 0.5346985459327698, 0.6546491980552673, 0.6114854216575623], [0.5509309768676758, 0.6506175398826599, 0.6896790266036987, 0.6683598160743713], [0.17333024740219116, 0.4564301073551178, 0.6126725077629089, 0.5369203090667725], [0.46935367584228516, 0.584055483341217, 0.6969013214111328, 0.6495816707611084], [0.35271865129470825, 0.47788241505622864, 0.5673419237136841, 0.5270097851753235], [0.2915976047515869, 0.48082032799720764, 0.6486634612083435, 0.5767485499382019], [0.5335884690284729, 0.5611255168914795, 0.6648507714271545, 0.6277702450752258], [0.4484841823577881, 0.5419958233833313, 0.7041314244270325, 0.6420139074325562], [0.3350512683391571, 0.499897837638855, 0.660699188709259, 0.593155026435852], [0.39079609513282776, 0.5599220991134644, 0.7641498446464539, 0.6816783547401428], [0.31215527653694153, 0.481784850358963, 0.6621860861778259, 0.587669849395752], [0.33140644431114197, 0.5579389333724976, 0.7648459076881409, 0.676747739315033], [0.4986553192138672, 0.5741947889328003, 0.695041298866272, 0.6482614278793335], [0.4856955409049988, 0.5386757850646973, 0.6581851840019226, 0.6137118339538574], [0.6038806438446045, 0.6272231340408325, 0.7211292386054993, 0.6876897215843201], [0.38944071531295776, 0.5182400941848755, 0.6805933713912964, 0.615496814250946], [0.2963932752609253, 0.3931616246700287, 0.609151303768158, 0.5286908149719238], [0.40463271737098694, 0.5516195893287659, 0.7317336201667786, 0.6591783761978149], [0.5658208131790161, 0.6982519030570984, 0.756440281867981, 0.7260050177574158], [0.42534375190734863, 0.5250095129013062, 0.6686892509460449, 0.6122592687606812], [0.46840324997901917, 0.5911034345626831, 0.6898998022079468, 0.6466368436813354], [0.4003022015094757, 0.5618664026260376, 0.7273980975151062, 0.6585366129875183]]}

def plot_precision_recall(history, title ='', num_epochs=40):
    import matplotlib.pyplot as plt

    prec = []
    rec = []
    # plot the precision and recall
    for i in range(num_epochs):
        prec.append(history['val_precision'][i][3])
        rec.append(history['val_recall'][i][3])
    plt.plot(prec, 'o-', label='Precision',)
    plt.plot(rec, 'o-',label='Recall')
    # put the legend out of the plot
    plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))
    plt.title('Precision and Recall: ' + title)
    plt.xlabel('Epoch')
    plt.ylabel('')
    # min max [0,1] yaxis
    plt.ylim([0.3, 1])
    # add grid
    plt.grid(True, alpha=0.2)
    plt.show()

def plot_precision_recall_list(list_of_history_dicts, suptitle=None, subplot_titles=None, num_epochs=40,  ):
    num_experiments = len(list_of_history_dicts)

    fig, axs = plt.subplots(1, num_experiments, figsize=(num_experiments * 6 + 2, 5),
                            sharex=False,
                            sharey=False)
    
    if num_experiments == 1:
        axs = np.expand_dims(axs, axis=0)

    for i in range(num_experiments):
        # Get values for each experiment
        history = list_of_history_dicts[i]
        prec = [sublist[3] for sublist in history['val_precision'] if len(sublist) > 3]
        rec = [sublist[3] for sublist in history['val_recall'] if len(sublist) > 3]
        prec = prec[:num_epochs]
        rec = rec[:num_epochs]
        epochs = range(1, num_epochs + 1)

        # Plot precision and recall
        axs[i].plot(epochs, prec, 'o-', label='Precision')
        axs[i].plot(epochs, rec, 'o-', label='Recall')
        axs[i].set_title(subplot_titles[i] if subplot_titles is not None else '', fontsize=20)
        axs[i].set_xlabel('Epoch')
        axs[i].set_ylabel('')
        axs[i].grid(True, alpha=0.2)
        # set min and max on y-axis to be [0,1]
        axs[i].set_ylim([0.3, 1])
        if i == (num_experiments-1):
            axs[i].legend(loc='center left', bbox_to_anchor=(1, 0.5), fontsize=18)
        
    if suptitle is not None:
        plt.suptitle(suptitle, fontsize=20)
    
    plt.show()    

plot_precision_recall_list([exp1_seed20_history_precision_recall,exp3_seed20_history_precision_recall], subplot_titles=['exp1 DLV3P Seed 20', 'exp3 PAN Seed 20'])
```

```python
# Plot precision-recall curve for the four decision threshold options
def get_max_val_iou_epoch(hist_list):
    max_val_iou_epoch = []
    for i in range(len(hist_list)):
        max_val_iou_epoch.append(np.argmax(hist_list[i]['val_iou']))
    # use the max_val_iou_epoch to get the precision and recall values
    max_val_iou_precision = []
    max_val_iou_recall = []
    for i in range(len(hist_list)):
        max_val_iou_precision.append(hist_list[i]['val_precision'][max_val_iou_epoch[i]][-1])
        max_val_iou_recall.append(hist_list[i]['val_recall'][max_val_iou_epoch[i]][-1])
    return max_val_iou_precision, max_val_iou_recall

decision_thresh = [0.3, 0.4, 0.5, 0.6]
decision_thresh_colors = ['r', 'g', 'b', 'y']
DLV3P_hist_list_thresh_exp = [exp8_6_hist, exp8_5_hist, exp8_hist, exp8_7_hist]
max_val_iou_precision_dlv3p, max_val_iou_recall_dlv3p = get_max_val_iou_epoch(DLV3P_hist_list_thresh_exp)
PAN_hist_list_thresh_exp = [exp9_6_hist, exp9_5_hist, exp9_hist, exp9_7_hist]
max_val_iou_precision_pan, max_val_iou_recall_pan = get_max_val_iou_epoch(PAN_hist_list_thresh_exp)

plt.figure()
# plot each threshold's precision and recall and label it by the thresh
for i in range(len(decision_thresh)):
    plt.scatter(max_val_iou_recall_dlv3p[i], max_val_iou_precision_dlv3p[i], label='DLV3P; threshold {}'.format(decision_thresh[i]), color=decision_thresh_colors[i])
    plt.scatter(max_val_iou_recall_pan[i], max_val_iou_precision_pan[i], label='PAN; threshold {}'.format(decision_thresh[i]), color=decision_thresh_colors[i], marker='x')

plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall curve for different decision thresholds')
plt.legend()
plt.grid()
plt.show()

```

```python
def plot_experiment_results(exp_nums, train_losses, val_ious, last_epochs, descriptions=None,
                            share_y_train_loss=False, share_y_val_iou=False):
    num_experiments = len(exp_nums)

    # If no descriptions are provided, create an empty description for each experiment
    if descriptions is None:
        descriptions = [''] * num_experiments

    # Determine whether to share y-axis across rows
    sharey = None
    if share_y_train_loss and share_y_val_iou:
        sharey = 'row'
    elif share_y_train_loss or share_y_val_iou:
        sharey = None  # To be controlled individually

    # Create a figure with subplots arranged in 2 rows and `num_experiments` columns
    fig, axs = plt.subplots(2, num_experiments, figsize=(num_experiments * 6, 8),
                            sharex=False,
                            sharey=sharey)

    # Ensure axs is a 2D array, even if there is only one experiment
    if num_experiments == 1:
        axs = np.expand_dims(axs, axis=1)

    for i in range(num_experiments):
        # Get values for each experiment
        exp_num = exp_nums[i]
        train_loss = train_losses[i]
        val_iou = val_ious[i]
        last_epoch = last_epochs[i]
        description = descriptions[i]

        # Calculate the epoch numbers for plotting
        epoch_nums = np.arange(last_epoch - len(train_loss), last_epoch)

        # Plot training loss in the first row
        axs[0, i].plot(epoch_nums, train_loss, label='Training Loss', marker='o')
        axs[0, i].set_title('Training Loss')
        axs[0, i].set_xlabel('Epoch')
        axs[0, i].set_ylabel('Loss')
        axs[0, i].grid(True, alpha=0.2)

        # Plot validation IoU in the second row
        axs[1, i].plot(epoch_nums, val_iou, label='Validation IoU', marker='o', color='orange')
        best_epoch_ind = np.argmax(val_iou)  # Mark the best IoU epoch
        axs[1, i].plot(epoch_nums[best_epoch_ind], val_iou[best_epoch_ind], marker='o', color='red', label='Best IoU')
        axs[1, i].set_title('Validation IoU')
        axs[1, i].set_xlabel('Epoch')
        axs[1, i].set_ylabel('Combined IoU')
        if i == num_experiments - 1: axs[1, i].legend()
        axs[1, i].grid(True, alpha=0.2)

        # Add experiment number and description as a header above each column
        fig.text(0.5 / num_experiments + i / num_experiments, 0.95, f'{exp_num}; {description}',
                 ha='center', va='center', fontsize=14, weight='bold')
    
    plt.tight_layout(rect=[0, 0, 1, 0.93])  # Adjust the layout to make room for the headers
    plt.show()
```
